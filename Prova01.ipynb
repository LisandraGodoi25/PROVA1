Creio que a estrutura de dados mais adequada para armazenar os chamados seja lista de dicionários talvez, pois os chamados são atendidos na ordem que chegam, mas caso algum chamado seja realizado por engano, ele deve voltar ao inicio da fila de uma forma que me recorda de uma pilha.

O  atendimento mais recente pode talvez ficar em um self.ultimo, só em caso de precisar ser readicionado.

Se a lista for 1 2 3 4 5 6
atender 2 3 4 5 6
atender 3 4 5 6
desfazer 2 3 4 5 6
atender 3 4 5 6

```python

class centralAtendimento():
    def __init__(self):
        self.atendimentos = {}
        self.atender = []
        self.atendido = []
        self.ultimo = ""

    def adicionar_chamado(self, codigo, cliente, prioridade):
        self.atendimentos[codigo] = {"cliente": cliente, "prioridade": prioridade}

        self.atender.append(codigo)

    def atender_proximo(self):
        if self.atender:
            self.ultimo = self.atender.pop(0)
            self.atendido.append(self.ultimo)

    def desfazer_ultimo_atendimento(self):
        if self.ultimo and self.ultimo in self.atendido:
            self.atendido.remove(self.ultimo)
            self.atender.insert(0, self.ultimo)
            
            self.ultimo = ""

    def buscar_chamado(self, codigo):
        if codigo in self.atendimentos:
            print(f"Cliente: {self.atendimentos[codigo]['cliente']}")
            print(f"Prioridade: {self.atendimentos[codigo]['prioridade']}")
        else:
            print("Chamado não encontrado!")

    def exibir_situacao(self):
        print("\nATENDER")
        for item in self.atender:
            print(item)

        print("\nATENDIDO")
        for item in self.atendido:
            print(item)
```

```python
central = centralAtendimento()

central.adicionar_chamado(101, "Ana", "Média")
central.adicionar_chamado(102, "Bruno", "Alta")
central.adicionar_chamado(103, "Carla", "Baixa")
central.adicionar_chamado(104, "Daniel", "Alta")

central.atender_proximo()
central.atender_proximo()

central.desfazer_ultimo_atendimento()

central.atender_proximo()

central.exibir_situacao()

central.buscar_chamado(104)
```

```python
print(central.buscar_chamado(104))
```
